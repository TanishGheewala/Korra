```python
from datetime import datetime, timezone
from pathlib import Path
import os
import time
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
from korra_m10.chat_graph import build_chat
from korra_m10.config import text_content

root = Path.cwd()
if root.name == "S1":
    root = root.parent
load_dotenv(root / ".env", override=False)
assert os.getenv("GOOGLE_API_KEY"), "Set GOOGLE_API_KEY in .env"
print("Executed:", datetime.now(timezone.utc).isoformat())
```

```python
memory = InMemorySaver()
graph = build_chat(memory)
thread_1 = {"configurable": {"thread_id": "notebook-1"}}
thread_2 = {"configurable": {"thread_id": "notebook-2"}}

def ask(text, config):
    response = graph.invoke({"messages": [HumanMessage(content=text)]}, config)
    print(text_content(response["messages"][-1]))
    return response
```

## One thread retains context across multiple turns


```python
ask("My name is Casey. Please remember it.", thread_1)
ask("My favorite color is turquoise. Remember that too.", thread_1)
remembered = ask("What is my name and favorite color?", thread_1)
print("Messages in thread 1:", len(graph.get_state(thread_1).values["messages"]))
```
